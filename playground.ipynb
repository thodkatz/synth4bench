We would like to build a classifier that given a vcf from one of that given callers, and after
post-processing it to filter the features found in our dataset, we would be able to verify/estimate
via the ML model the credibility of the variant findings.

1. First filter the models to include the records that included caller information. Thus classes in TN and FN aren't relevant because callers output only the variants that they spotted, so for the negative case, absense of variant, there isn't information of the callers in the first place.
2. Based on neomutate feature selection, we can be based on what to use.

Features for our xdgboost approach:
- is_snv_indel
- is_transcrption_tranversion
- Coverage
- Read length
- DP
- AF
- Caller

POS, REF, ALT are dropped. POSITION is volatile, references at test time will be different, won't be a valuable info. REF, ALT information is encoded via is_transcription_transversion based on neomutate. Similalry categoration of SNV, Indel, is integrated with is_snv_indel.

What would be the benefit to encode within one variable all the transversion, transition, insertion, delettion, instead of splitting in two broad categories, one snv_indel and the other identifying insertion, deletcion, transition or transversion based on the parent category. In our approach since the caller will be a decision making node, effectively the model won't be able to leverage information across model. It would be able to detect per caller, the vulnerabilities that it may have across the different features that will be used. For example it could detect that for specific coverage, af could lead to a less confident evaluation. An ensemble ml approach would be able to leverage the cross caller functionality to increase the accuracy, but not estimating the confidence of the prediction of the caller.
